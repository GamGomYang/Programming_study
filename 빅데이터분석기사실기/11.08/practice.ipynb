```python
import pandas as pd

train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch5/train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch5/test.csv")

train.info()

print(train.head())

target = train.pop('price')
train = pd.get_dummies(train)
test = pd.get_dummies(test)

from sklearn.model_selection import train_test_split

x_train , x_val , y_train , y_val = train_test_split(train,target,test_size = 0.2 ,random_state=0)


from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state = 0)

pred = model.fit(x_train, y_train)

from  sklearn.metrics import root_mean_squared_error

pred = model.predict(x_val)

print(root_mean_squared_error(y_val , pred))



```

```python
import pandas as pd
# df = pd.read_csv("data6-1-1.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch6/data6-1-1.csv")


df['출발'] = pd.to_datetime(df['출동시간'])
df['도착'] = pd.to_datetime(df['도착시간'])

df['차이'] = (df['도착']-df['출발']).dt.total_seconds()/60

print(df['차이'].head(3))

average = df.groupby('소방서')['차이'].mean()
print(average)
idx = average.idxmax()
result = average[idx]
print(round(result))
```

```python
import pandas as pd
# df = pd.read_csv("data6-1-2.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch6/data6-1-2.csv")

df['총학생'] = df['1학년']+df['2학년']+df['3학년']+df['4학년']+df['5학년']+df['6학년']

print(df.head(3))

df['비율'] = df['총학생']/df['교사수']

print(df['비율'].head())

df = df.sort_values('비율' , ascending = False)


print(df.iloc[0,1])
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch6/data6-1-3.csv")

df.info()

print(df.head(3))

df['총범죄'] = (df['강력범죄'] + df['절도범죄'] + df['지능범죄']+ df['풍속범죄']+df['교통범죄'])


df['연도'] = df["날짜"].str[:4]

print(df['연도'].head(3))

result = df['총범죄'].groupby(df['연도']).sum()

print(round(result.max()/12))
```

```python
# 데이터 불러오기
import pandas as pd
# train = pd.read_csv("energy_train.csv")
# test = pd.read_csv("energy_test.csv")
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch6/energy_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch6/energy_test.csv")

train.info()

print(train.head(3))

target = train.pop('Heat_Load')

train = pd.get_dummies(train)

test = pd.get_dummies(test)


from sklearn.model_selection import train_test_split

x_train , x_val , y_train , y_val = train_test_split(train , target , test_size = 0.2 ,random_state = 0)

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import f1_score

model = RandomForestClassifier(random_state = 0)

pred = model.fit(x_train , y_train)


pred = model.predict(x_val)

print(f1_score(y_val , pred, average = 'macro'))
```
