```python
import pandas as pd
df = pd.DataFrame({
    '충전기': ['New'] * 10 + ['Old'] * 10,
    '충전시간': [
        1.5, 1.6, 1.4, 1.7, 1.5, 1.6, 1.7, 1.4, 1.6, 1.5,
        1.7, 1.8, 1.7, 1.9, 1.8, 1.7, 1.8, 1.9, 1.7, 1.6
    ]
})

print(df.head(2))
```

```python
from scipy import stats

new_cond = df['충전기'] == 'New'
old_cond = df['충전기'] == 'Old'

print(stats.ttest_ind(df[new_cond]['충전시간'] , df[old_cond]['충전시간'],alternative = 'less', equal_var = True))
```
