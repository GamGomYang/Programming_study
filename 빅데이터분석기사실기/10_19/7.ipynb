```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1~3. 독립성 검정
observed = pd.DataFrame([[50, 30], [60, 40]])
print(chi2_contingency(observed))
```

```python
import pandas as pd
df = pd.DataFrame({
    '캠프' : ['빅분기']*80 + ['정처기']*100,
    '등록여부' : ['등록']*50 + ['등록안함']*30+ ['등록']*60 + ['등록안함']*40

})
print(df.head())
```

```python
df  = pd.crosstab(df['캠프'], df['등록여부'])
print(df)

print(chi2_contingency(df))
```
