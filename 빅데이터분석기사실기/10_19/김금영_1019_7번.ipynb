# SECTION 07. 독립성 검정 (Chi-Square Test of Independence)


길벗 대학에서 빅데이터 분석기사 캠프와 정보처리기사 캠프가 있다.  
두 학생들 사이에 주최한 세미나에 참석하려면 등록을 해야 한다.  
각 캠프의 세미나 등록 여부 를 조사한 결과는 아래와 같다.

| 빅데이터 분석기사  50  30 |
| 정보처리기사       60  40 |

캠프와 세미나 등록 여부가 서로 독립적인지 검정
(유의수준 0.05)


```python
import pandas as pd
from scipy.stats import chi2_contingency

#카이제곱(χ²) 독립성 검정 -> 두 범주형 변수 사이에 통계적으로 유의한 연관이 있는지 확인
observed = pd.DataFrame([[50, 30], [60, 40]])
print(chi2_contingency(observed))

Chi2ContingencyResult(statistic=0.03535714285714309, pvalue=0.8508492527705047, dof=1, expected_freq=array([[48.88888889, 31.11111111],
[61.11111111, 38.88888889]]))

# chi2 : 계산된 χ² 통계량
# p : p-value (귀무가설을 기각할 확률)
# dof : 자유도 = (행 개수 - 1) × (열 개수 - 1)
# expected : 기대도수 표 -> 두 변수가 진짜로 독립이라면 각 칸에 예상되는 값
```

```python
import pandas as pd
#데이터 생성
df = pd.DataFrame({
    '캠프' : ['빅분기']*80 + ['정처기']*100,
    '등록여부' : ['등록']*50 + ['등록안함']*30+ ['등록']*60 + ['등록안함']*40

})
print(df.head())
```

```python
#범주형 변수 간의 교차 빈도표 생성
df  = pd.crosstab(df['캠프'], df['등록여부'])
print(df)


#카이제곱 독립성 검정 실행
print(chi2_contingency(df))


# 등록여부  등록  등록안함
# 캠프            
# 빅분기   50    30
# 정처기   60    40
# Chi2ContingencyResult(statistic=0.03535714285714309, pvalue=0.8508492527705047, dof=1, expected_freq=array([[48.88888889, 31.11111111],
#        [61.11111111, 38.88888889]]))
```
