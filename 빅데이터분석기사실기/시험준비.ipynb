```python
# 데이터 불러오기
import pandas as pd
# train = pd.read_csv("train.csv")
# test = pd.read_csv("test.csv")
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch3/train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch3/test.csv")

print(train.head(3))
print(train.info())

# Employment GraduateOrNot

from sklearn.preprocessing import LabelEncoder


le = LabelEncoder()

train.shape , test.shape

# train = le.fit_transform(train)
# test = le.transform(test)

train = pd.get_dummies(train)
test = pd.get_dummies(test)
target = train.pop('TravelInsurance')

from sklearn.model_selection import train_test_split

x_train, x_val , y_train , y_val = train_test_split(train, target , test_size = 0.2 , random_state = 0)

from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(random_state = 0)

model.fit(x_train, y_train)
pred = model.predict(x_val)


from sklearn.metrics import f1_score
f1score = f1_score(pred, y_val)

print(f1score)


#1. 0.77 (random + onehot encoding)


result = pd.DataFrame({'pred':pred})
result.to_csv("submit.csv")
```

```python
import pandas as pd
# df = pd.read_csv("clam.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/clam.csv")

# from statsmodels.formula.api import logit 
# import numpy as np
train = df[:210]
test = df[210:]

# print(train.shape) 
# print(test.shape)


# formula = 'gender ~ weight '

# model = logit(formula , data = train).fit()

# odds_ratio = np.exp(model.params['weight'])
# print(round(odds_ratio,4))

from statsmodels.formula.api import logit
import numpy as np
formula = 'gender ~ age + length + diameter + height + weight'
# formula 독립변수 설정시 +로 지정 fomula = 'gender ~ age + length'

model = logit(formula , data = train).fit()

print(round(-2*model.llf,2))
#잔차 이탈도 - 로그우도에 -2 곱한것 -2*model.llf


#error rate = 1- accuracy

from sklearn.metrics import accuracy_score
formula = 'gender ~ weight'
model = logit(formula, data = train).fit()
target = test.pop('gender')
pred= model.predict(test)
pred = (pred>0.5).astype(int)

accuracy = accuracy_score(target, pred)
error_rate = 1-accuracy
print(round(error_rate,3))


```
