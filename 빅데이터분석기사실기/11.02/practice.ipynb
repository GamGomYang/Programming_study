```python
import pandas as pd

train = pd.read_csv('customer_train.csv')
test = pd.read_csv('customer_test.csv')

#print(train.head(10))

#preprocessing
#null preprocessing

train['payback'] = train['payback'].fillna(0)
test['payback'] = test['payback'].fillna(0)


print(train.head(5))

from sklearn.preprocessing import LabelEncoder

#label encoding

cols = ['main_product','where']

for col in cols:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    test[col] = le.fit_transform(test[col])


target = train.pop('gender')

#data split

from sklearn.model_selection import train_test_split

x_train , x_val , y_train, y_val = train_test_split(train, target, test_size = 0.2 , random_state =0)
# why zero value int land state?


# ml 

from sklearn.ensemble  import RandomForestClassifier

model = RandomForestClassifier(random_state = 0)
model.fit(x_train, y_train)


from sklearn.metrics import roc_auc_score

pred = model.predict_proba(x_val)

print(roc_auc_score(y_val, pred[:,1]))


pred = model.predict_proba(test)
submit = pd.DataFrame({'pred': pred[:,1]})
submit.to_csv('submission.csv', index = False)
print(submit.head())

```

```python
import pandas as pd


train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/test.csv")


target = train.pop('Segmentation')

print(train.head(5))

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder
train.describe()

train = pd.get_dummies(train)
test = pd.get_dummies(test)

print(train.shape , test.shape)

from sklearn.model_selection import train_test_split

x_train , x_val , y_train, y_val = train_test_split(train, target , test_size = 0.2 , random_state = 0)

from sklearn.ensemble  import RandomForestClassifier

model = RandomForestClassifier(random_state = 0)

model.fit(x_train, y_train)

pred = model.predict(x_val)

from sklearn.metrics import f1_score

print(f1_score(y_val, pred, average='macro'))

pred = model.predict(test)

submit = pd.DataFrame({
    'ID': test['ID'],
    'Segmentation': pred
})
submit.to_csv("result.csv",index = False)

pd.read_csv("result.csv")

```
