```python
import pandas as pd

df_train = pd.read_csv('customer_train.csv')
df_test = pd.read_csv('customer_test.csv')

#df_train.info()


# preprocessing - filled NaN data with 0 
df_train['payback'] = df_train['payback'].fillna(0)
df_test['payback'] = df_test['payback'].fillna(0)
#print(df_train['payback'].head())
#print(df_test['payback'].head())

#label encoding
#sklearn.preprocessing - LabelEncoder(Pascal) - transform object into integer
# ml(sckit learn) only needs integer value
from sklearn.preprocessing import LabelEncoder

cols = ['main_product','where']

for col in cols:
    le = LabelEncoder()
    df_train[col] = le.fit_transform(df_train[col])
    df_test[col] = le.fit_transform(df_test[col])


target = df_train.pop('gender')

#split test, train cases

from sklearn.model_selection import train_test_split

X_tr, X_val , Y_tr, Y_val = train_test_split(df_train, target , test_size = 0.2, random_state = 0)
print(X_tr.shape, X_val.shape, Y_tr.shape , Y_val.shape)


#ML process
#ensemble - randomforest 
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state = 0)
model.fit(X_tr , Y_tr)
pred = model.predict_proba(X_val)
print(pred)

from sklearn.metrics import roc_auc_score
print(roc_auc_score(Y_val ,pred[:,1]))

pred = model.predict_proba(df_test)

submit = pd.DataFrame({
    'pred' : pred[:,1]
})

print(submit.head(3))
```

```python
# -----------------------------
# 🧩 1. Titanic 데이터 불러오기
# -----------------------------
import pandas as pd

# 타이타닉 생존자 데이터를 GitHub에서 불러옴 (CSV 형식)
df = pd.read_csv('https://raw.githubusercontent.com/visualpython/visualpython/main/visualpython/data/sample_csv/titanic.csv')

#print(df.head())  # 데이터 앞부분 확인용 (주석처리됨)

print("1번")

# -----------------------------
# 📊 2. 성별(Sex)과 생존여부(Survived)의 교차표(crosstab) 생성
# -----------------------------
# 행: 성별, 열: 생존여부(0=사망, 1=생존)
crosstab = pd.crosstab(df['Sex'], df['Survived'])
print(crosstab)


# -----------------------------
# 🔍 3. 카이제곱 독립성 검정 (chi-square test of independence)
# -----------------------------
from scipy.stats import chi2_contingency

# crosstab을 이용해 두 변수(Sex, Survived)가 독립인지 검정
print(chi2_contingency(crosstab))
# 반환값: (카이제곱 통계량, p-value, 자유도, 기대빈도행렬)
# 예) (260.717..., 1.197e-58, 1, array([[여성기대값], [남성기대값]]))

# → p-value가 매우 작으면, 성별과 생존여부는 **독립이 아니다** (즉, 성별에 따라 생존 확률이 다름)

#result = 260.717  # 참고용 주석

print("2번")

# -----------------------------
# 📋 4. 데이터 정보 확인
# -----------------------------
df.info()  # 열 이름, 자료형, 결측값 여부 등 표시

# 모델에 사용할 주요 변수 일부 미리 확인
df[['Sex', 'SibSp', 'Parch', 'Fare']].head(3)

# -----------------------------
# 🧮 5. 로지스틱 회귀(Logistic Regression) 분석
# -----------------------------
from statsmodels.formula.api import logit

# x: 독립변수들(Sex, SibSp, Parch, Fare)
# y: 종속변수(Survived)
x = df[['Sex', 'SibSp', 'Parch', 'Fare']].copy()
y = df['Survived']

# 로지스틱 회귀모델 적합
# formula 형식: '결과변수 ~ 독립변수1 + 독립변수2 + ...'
model = logit(formula = "Survived ~ SibSp + Parch + Fare + Sex", data = df).fit()

# 모델 요약 출력
print(model.summary())
# → 각 변수별 회귀계수(coef), 표준오차, z값, p-value 등 제공
# p<0.05이면 통계적으로 유의한 변수로 볼 수 있음

# 임의로 -0.2007을 반올림 (예시로 사용된 값)
print(round(-0.2007,3))  # 출력: -0.201

print("3번")

# -----------------------------
# 📈 6. 회귀계수를 이용한 오즈비(Odds Ratio) 계산
# -----------------------------
import numpy as np

# "SibSp" (같이 탄 형제자매 수) 변수의 회귀계수를 지수화하여 오즈비 계산
odds_ratio = np.exp(model.params["SibSp"])
print(odds_ratio)

# 반올림하여 출력
print(round(odds_ratio,3))
# → 오즈비가 1보다 크면 형제자매 수가 늘수록 생존확률이 높아짐,
#   1보다 작으면 생존확률이 낮아짐을 의미함.
```

```python
import pandas as pd

# 데이터 불러오기 (학습/테스트 특성, 학습 타깃)
X_test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch2/X_test.csv")
X_train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch2/X_train.csv")
y_train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch2/y_train.csv")


from sklearn.preprocessing import LabelEncoder
# 범주형(object) 컬럼만 선택
cols = X_train.select_dtypes(include='object').columns

# 변환 전 샘플 확인
display(X_train[cols].head(3))

# 각 범주형 컬럼을 숫자로 변환(레이블 인코딩)
# fit_transform은 학습 데이터에 대해 학습+변환, transform은 동일한 매핑으로 테스트 데이터 변환
for col in cols:
    le = LabelEncoder()
    X_train[col] = le.fit_transform(X_train[col])
    X_test[col] = le.transform(X_test[col])

# 변환 후 샘플 확인
display(X_train[cols].head(3))

# ID 컬럼은 모델 입력에서 제거(식별용이므로)
X_train = X_train.drop('ID', axis=1)
# 테스트 데이터의 ID는 제출용으로 따로 보관
X_test_id = X_test.pop('ID')



from sklearn.model_selection import train_test_split
# 학습 데이터를 학습/검증으로 분리 (검증 비율 20%, 재현을 위해 random_state 고정)
X_tr,X_val,y_tr,y_val = train_test_split(X_train,y_train['Reached.on.Time_Y.N'],test_size=0.2,random_state=0)
X_tr.shape, X_val.shape, y_tr.shape, y_val.shape


# 필요한 모델 불러오기 (xgboost/lightgbm 사용 시 별도 설치 필요)
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb 
import lightgbm as lgb  
from sklearn.metrics import roc_auc_score


# 1) 로지스틱 회귀: 학습 후 검증 AUC 출력
lr = LogisticRegression(random_state=0)
lr.fit(X_tr, y_tr)
pred = lr.predict_proba(X_val)
print("Logistic AUC:", roc_auc_score(y_val, pred[:,1]))


# 2) 의사결정나무: 학습 후 검증 AUC 출력
dt = DecisionTreeClassifier(random_state=0)
dt.fit(X_tr, y_tr)
pred = dt.predict_proba(X_val)
print("DecisionTree AUC:", roc_auc_score(y_val, pred[:,1]))

# 3) 랜덤포레스트: 학습 후 검증 AUC 출력
rf = RandomForestClassifier(random_state=0)
rf.fit(X_tr, y_tr)
pred = rf.predict_proba(X_val)
print("RandomForest AUC:", roc_auc_score(y_val, pred[:,1]))


# 4) XGBoost: 학습 후 검증 AUC 출력
xg = xgb.XGBClassifier(random_state=0)
xg.fit(X_tr, y_tr)
pred = xg.predict_proba(X_val)
print("XGBoost AUC:", roc_auc_score(y_val, pred[:,1]))
```

```python
import pandas as pd


train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/test.csv")



target = train.pop('Segmentation')

# one-hot encoding
print(train.shape, test.shape)
train = pd.get_dummies(train)
test = pd.get_dummies(test)
print(train.shape, test.shape)


from sklearn.model_selection import train_test_split
X_tr, X_val, y_tr, y_val = train_test_split(train,target, test_size=0.2,random_state=0)
X_tr.shape, X_val.shape, y_tr.shape, y_val.shape

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
import lightgbm as lgb
from sklearn.metrics import f1_score

# 2 methode - logistic, lgb 
lr = LogisticRegression(random_state=0)
lr.fit(X_tr, y_tr)
pred = lr.predict(X_val)
print(f1_score(y_val, pred, average='macro'))

lg = lgb.LGBMClassifier(random_state=0, verbose=-1)
lg.fit(X_tr, y_tr)
pred = lg.predict(X_val)
print(f1_score(y_val, pred, average='macro'))

pred = lg.predict(test)
submit = pd.DataFrame({
    'ID': test['ID'],
    'Segmentation': pred
})
submit.to_csv("result.csv", index=False)

pd.read_csv("result.csv")
```
