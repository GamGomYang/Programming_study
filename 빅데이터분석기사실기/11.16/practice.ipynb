```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/chem.csv")
df.head()

scaler = MinMaxScaler()
df['co_scaler'] = scaler.fit_transform(df[['co']])
df['nmhc_scaler'] = scaler.fit_transform(df[['nmhc']])


co_std = df['co_scaler'].std()
nmhc_std = df['nmhc_scaler'].std()

print(co_std, nmhc_std)

std_diff = round(co_std - nmhc_std,3)

print(std_diff)
```

```python
import pandas as pd
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/churn_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/churn_test.csv")
train.head()
train.info()

#MAE - 회귀
#예측값이 연속값 = 회귀

train.isnull().sum()

train['TotalCharges'].describe()

target = train.pop('TotalCharges')

from sklearn.preprocessing import LabelEncoder

cols = train.select_dtypes(include = 'object').columns

for col in cols:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    test[col] = le.fit(test[col])

from sklearn.model_selection import train_test_split

x_train, x_val , y_train , y_val = train_test_split(train, target, test_size = 0.2 , random_state= 0)


from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state = 0)
rf.fit(x_train, y_train)

from sklearn.metrics import mean_absolute_error
pred = rf.predict(x_val)
print(mean_absolute_error(y_val, pred))

```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/churn.csv")

formula = "Churn ~ AccountWeeks + ContractRenewal + DataPlan + DataUsage + CustServCalls + DayMins + DayCalls + MonthlyCharge + OverageFee + RoamMins"


from statsmodels.formula.api import logit
model = logit(formula , data = df).fit()


sum(model.pvalues[1:]>0.05)


formula = "Churn~DataUsage + DayMins"

model = logit(formula, data = df).fit()

print(model.summary())

round(sum(model.params),3)

import numpy as np

coed = model.params['DataUsage']
round(np.exp(coed*5),3)
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/piq.csv")

from statsmodels.formula.api import ols

model = ols('PIQ~Brain+Height + Weight ', data=df).fit()

print(model.summary())

print(round(model.rsquared,2))

new_data = pd.DataFrame({'Brain':[90],'Height':[70],'Weight':[150]})
pred = model.predict(new_data)
print(pred)
print(round(pred[0]))
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/loan.csv")

df.head()

#총대출액 = 신용대출 + 담보대출

df['총대출액'] = df['신용대출']+df['담보대출']

df.info()
# 성별별 총대출액
groups = df.groupby(['지역코드','성별'])['총대출액'].sum().unstack()

# 설별간 총대출액의 차이가 가장 큰 지역코드
groups['차이'] = abs(groups[1]-groups[2])
result = groups['차이'].idxmax()
print(result)
```
